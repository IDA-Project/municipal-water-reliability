```python
# =====================================================================
# QUESTION 6.1 - CONTROLLED ABLATION STUDY
# =====================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    RepeatedStratifiedKFold,
    cross_validate
)

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LogisticRegression


SEED = 42


# =====================================================================
# 1. LOAD DATA
# =====================================================================

PATH = r"C:\Users\Sekomane\Desktop\Data ScienceTUT\NIWIS_Water Supply Reliability - Population_26-Sep-2026.csv"

df = pd.read_csv(PATH)

print("=" * 80)
print("QUESTION 6.1 - CONTROLLED ABLATION STUDY")
print("=" * 80)

print("Dataset shape:", df.shape)


# =====================================================================
# 2. CLEAN DATA
# =====================================================================

data = df.copy()

data = data.loc[
    :,
    ~data.columns.str.contains("^Unnamed")
]

data.columns = (
    data.columns
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
)


def clean_number(series):

    return pd.to_numeric(
        series.astype(str)
              .str.replace(" ", "", regex=False)
              .str.replace(",", "", regex=False)
              .str.strip(),
        errors="coerce"
    )


for col in data.columns:

    if "%" not in col and col != "WSA Name":

        data[col] = clean_number(
            data[col]
        )


for col in data.columns:

    if "%" in col:

        data[col] = pd.to_numeric(
            data[col]
                .astype(str)
                .str.replace("%", "", regex=False)
                .str.strip(),
            errors="coerce"
        )


# =====================================================================
# 3. CREATE TARGET
# =====================================================================

data["low_reliability"] = (
    data["% Population Reliable Supply"] < 60
).astype(int)


# =====================================================================
# 4. REMOVE LEAKAGE VARIABLES
# =====================================================================

leakage_columns = [

    "Population Reliable Supply",
    "% Population Reliable Supply",

    "Urban Population Reliable Supply",
    "% Urban Population Reliable Supply",

    "Rural Population Reliable Supply",
    "% Rural Population Reliable Supply"
]


y = data["low_reliability"]


X_base = data.drop(
    columns=[
        "low_reliability",
        "WSA Name"
    ] + leakage_columns,
    errors="ignore"
).copy()


# =====================================================================
# 5. CREATE FULL FEATURE SET
# =====================================================================

X_full = X_base.copy()


X_full["population_access_rate"] = (
    X_full["Population with access to water"]
    / X_full["Population"]
) * 100


X_full["urban_access_rate"] = (
    X_full["Urban Population with access to water"]
    / X_full["Urban Population"]
) * 100


X_full["rural_access_rate"] = (
    X_full["Rural Population with access to water"]
    / X_full["Rural Population"]
) * 100


X_full["urban_population_share"] = (
    X_full["Urban Population"]
    / X_full["Population"]
) * 100


X_full["rural_population_share"] = (
    X_full["Rural Population"]
    / X_full["Population"]
) * 100


X_full = X_full.replace(
    [np.inf, -np.inf],
    np.nan
)


X_base = X_base.replace(
    [np.inf, -np.inf],
    np.nan
)


# =====================================================================
# 6. CREATE SAME DEVELOPMENT / TEST INDICES
# =====================================================================

# We split using indices so the EXACT SAME observations are used
# for every ablation.

indices = np.arange(
    len(data)
)


dev_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=SEED,
    stratify=y
)


X_full_dev = X_full.iloc[
    dev_idx
].copy()


X_base_dev = X_base.iloc[
    dev_idx
].copy()


y_dev = y.iloc[
    dev_idx
].copy()


print("\nDevelopment observations:", len(dev_idx))
print("Held-out observations    :", len(test_idx))

print(
    "\nThe held-out test cases are not used to choose "
    "the ablation configuration."
)


# =====================================================================
# 7. CROSS-VALIDATION PROTOCOL
# =====================================================================

cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=3,
    random_state=SEED
)


scoring = {

    "accuracy":
        "accuracy",

    "macro_f1":
        "f1_macro",

    "weighted_f1":
        "f1_weighted",

    "roc_auc":
        "roc_auc",

    "pr_auc":
        "average_precision"
}


# =====================================================================
# 8. FULL OPTIMISED CHAMPION
# =====================================================================

full_model = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median",
            add_indicator=True
        )
    ),

    (
        "scaler",
        RobustScaler()
    ),

    (
        "classifier",
        LogisticRegression(
            C=1,
            penalty="l1",
            solver="liblinear",
            class_weight=None,
            max_iter=5000,
            random_state=SEED
        )
    )
])


# =====================================================================
# 9. ABLATION 1 - REMOVE ENGINEERED FEATURES
# =====================================================================

no_engineered_features_model = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median",
            add_indicator=True
        )
    ),

    (
        "scaler",
        RobustScaler()
    ),

    (
        "classifier",
        LogisticRegression(
            C=1,
            penalty="l1",
            solver="liblinear",
            class_weight=None,
            max_iter=5000,
            random_state=SEED
        )
    )
])


# =====================================================================
# 10. ABLATION 2 - REMOVE ROBUST SCALING
# =====================================================================

no_scaling_model = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median",
            add_indicator=True
        )
    ),

    (
        "classifier",
        LogisticRegression(
            C=1,
            penalty="l1",
            solver="liblinear",
            class_weight=None,
            max_iter=5000,
            random_state=SEED
        )
    )
])


# =====================================================================
# 11. ABLATION 3 - CHANGE L1 TO L2
# =====================================================================

l2_model = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median",
            add_indicator=True
        )
    ),

    (
        "scaler",
        RobustScaler()
    ),

    (
        "classifier",
        LogisticRegression(
            C=1,
            penalty="l2",
            solver="liblinear",
            class_weight=None,
            max_iter=5000,
            random_state=SEED
        )
    )
])


# =====================================================================
# 12. DEFINE EXPERIMENTS
# =====================================================================

experiments = [

    (
        "Full Optimised Model",
        full_model,
        X_full_dev
    ),

    (
        "Ablation 1: No Engineered Features",
        no_engineered_features_model,
        X_base_dev
    ),

    (
        "Ablation 2: No RobustScaler",
        no_scaling_model,
        X_full_dev
    ),

    (
        "Ablation 3: L2 instead of L1",
        l2_model,
        X_full_dev
    )
]


# =====================================================================
# 13. RUN CONTROLLED ABLATIONS
# =====================================================================

results = []


print("\n" + "=" * 80)
print("RUNNING CONTROLLED ABLATIONS")
print("=" * 80)


for experiment_name, model, X_current in experiments:

    print(
        f"\nEvaluating: {experiment_name}"
    )


    scores = cross_validate(
        model,
        X_current,
        y_dev,
        cv=cv,
        scoring=scoring,
        n_jobs=1
    )


    results.append({

        "Experiment":
            experiment_name,

        "Accuracy Mean":
            scores["test_accuracy"].mean(),

        "Macro-F1 Mean":
            scores["test_macro_f1"].mean(),

        "Macro-F1 SD":
            scores["test_macro_f1"].std(),

        "Weighted-F1 Mean":
            scores["test_weighted_f1"].mean(),

        "ROC-AUC Mean":
            scores["test_roc_auc"].mean(),

        "PR-AUC Mean":
            scores["test_pr_auc"].mean()
    })


# =====================================================================
# 14. RESULTS TABLE
# =====================================================================

results_df = pd.DataFrame(
    results
)


# Full-model Macro-F1
full_macro_f1 = results_df.loc[
    results_df["Experiment"] == "Full Optimised Model",
    "Macro-F1 Mean"
].iloc[0]


# Change relative to full model
results_df[
    "Macro-F1 Change vs Full"
] = (
    results_df["Macro-F1 Mean"]
    - full_macro_f1
)


numeric_columns = results_df.select_dtypes(
    include=np.number
).columns


results_df[numeric_columns] = (
    results_df[numeric_columns]
        .round(3)
)


print("\n" + "=" * 100)
print("QUESTION 6.1 - ABLATION RESULTS")
print("=" * 100)


display(
    results_df
)


# =====================================================================
# 15. INDIVIDUAL PERFORMANCE CHANGES
# =====================================================================

print("\n" + "=" * 80)
print("MACRO-F1 CHANGES")
print("=" * 80)


for _, row in results_df.iterrows():

    if row["Experiment"] != "Full Optimised Model":

        print(
            row["Experiment"]
        )

        print(
            "Macro-F1:",
            row["Macro-F1 Mean"]
        )

        print(
            "Change vs full model:",
            row["Macro-F1 Change vs Full"]
        )

        print()


# =====================================================================
# 16. ABLATION BAR CHART
# =====================================================================

plt.figure(
    figsize=(11, 6)
)


plt.bar(
    results_df["Experiment"],
    results_df["Macro-F1 Mean"],
    yerr=results_df["Macro-F1 SD"],
    capsize=6
)


plt.ylabel(
    "Mean Cross-Validation Macro-F1"
)


plt.xlabel(
    "Experiment"
)


plt.title(
    "Controlled Ablation Study"
)


plt.xticks(
    rotation=20,
    ha="right"
)


plt.ylim(
    0,
    1
)


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()

plt.show()


# =====================================================================
# 17. COMPLETION MESSAGE
# =====================================================================

print("\n" + "=" * 80)

print(
    "QUESTION 6.1 COMPLETED SUCCESSFULLY"
)

print("=" * 80)


print(
    "\nEach ablation changed only one component while "
    "the split, random seed, model family and evaluation "
    "protocol were kept fixed."
)
```
